```python
# Scaling the datas, meaning replacing each adata.X values with the z values and maximum z value it will keep it 10. where z = (value - mean)/standard_deviation.
sc.pp.scale(adata, max_value=10)
```

```python
# finding highly pca variance ratio,here we are not taking beyond 29. the 2000 columns we have is converted into 50 dimensions.
sc.pl.pca_variance_ratio(adata, n_pcs=50)
```

```python
# from 50 dimensions we only keep 29 highly important dimensions. now we create a KNN GRAPH, n_neighbour cluster 20 cells together(Group 20 cells together)
# This creates graph networks, n_neighbour create we from 20 closet cell from each cells and that creates a massive interconnected web.
sc.pp.neighbors(adata, n_neighbors=20, n_pcs=24)
```

```python
# the KNN algorithm creates graph but with this leiden we create a cluster.
# the linden algorithm creates clusters from the graph.
sc.tl.leiden(adata, resolution = 2.0)
```

```python
sc.tl.umap(adata)
```

```python
sc.pl.umap(adata, color='leiden')
```

```python
# Finding marker genes from each clusters.
# method = 'wilcoxon This is a statistical test. It looks at Cluster 0 and compares its RNA to all other clusters. If Gene A is super high in Cluster 0 but low everywhere else, it flags Gene A as a "Marker Gene" for Cluster 0.'
sc.tl.rank_genes_groups(adata, groupby = 'leiden', method = 'wilcoxon')
```
